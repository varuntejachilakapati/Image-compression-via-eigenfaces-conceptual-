```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_olivetti_faces

# Standard dataset: Olivetti Faces
faces = fetch_olivetti_faces(shuffle=False, random_state=0)

# Take 5 faces
X = faces.data[:5]          # (5, 4096)
images = X.reshape(5, 64, 64)

# -----------------------------
# 1. Mean face
# -----------------------------
mean_face = X.mean(axis=0)

# Center the faces
X_centered = X - mean_face

# -----------------------------
# 2. Covariance matrix
# -----------------------------
C = (X_centered.T @ X_centered) / (len(X) - 1)

print("Face matrix:", X.shape)
print("Covariance matrix:", C.shape)

# -----------------------------
# 3. Eigenvalues/eigenvectors
# -----------------------------
eigenvalues, eigenvectors = np.linalg.eigh(C)

# Sort from largest to smallest
idx = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

# Top 3 eigenfaces
eigenfaces = eigenvectors[:, :3]

print("\nTop 3 eigenvalues:")
for i in range(3):
    print(f"lambda_{i+1} = {eigenvalues[i]:.8f}")

# -----------------------------
# 4. Project faces onto
#    3-dimensional eigenspace
# -----------------------------
coefficients = X_centered @ eigenfaces

# -----------------------------
# 5. Reconstruct
# -----------------------------
X_reconstructed = (
    mean_face +
    coefficients @ eigenfaces.T
)

# Reconstruction error
rmse = np.sqrt(
    np.mean((X - X_reconstructed)**2, axis=1)
)

print("\nReconstruction RMSE:")
for i, error in enumerate(rmse):
    print(f"Face {i+1}: {error:.6f}")

# -----------------------------
# 6. Show original vs reconstructed
# -----------------------------
fig, axes = plt.subplots(2, 5, figsize=(12, 5))

for i in range(5):
    axes[0, i].imshow(images[i], cmap="gray")
    axes[0, i].set_title(f"Original {i+1}")
    axes[0, i].axis("off")

    axes[1, i].imshow(
        X_reconstructed[i].reshape(64, 64),
        cmap="gray"
    )
    axes[1, i].set_title(f"3-EF Recon {i+1}")
    axes[1, i].axis("off")

plt.tight_layout()
plt.show()

# -----------------------------
# 7. Display eigenfaces
# -----------------------------
fig, axes = plt.subplots(1, 3, figsize=(9, 3))

for i in range(3):
    axes[i].imshow(
        eigenfaces[:, i].reshape(64, 64),
        cmap="gray"
    )
    axes[i].set_title(f"Eigenface {i+1}")
    axes[i].axis("off")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from sklearn.decomposition import PCA
import gradio as gr
import io

# ============================================================
# BACKEND: EIGENFACE / PCA IMAGE COMPRESSION
# ============================================================

def compress_image(image, n_components):
    """
    Compress an image using PCA (Eigenfaces concept).
    """

    # Convert image to grayscale
    img = Image.fromarray(image).convert("L")

    # Resize for faster processing
    max_size = 512
    img.thumbnail((max_size, max_size))

    img_array = np.array(img, dtype=np.float64)

    # PCA works on rows of the image
    # Each row is treated as a sample
    n_components = min(
        int(n_components),
        img_array.shape[0],
        img_array.shape[1]
    )

    # Apply PCA
    pca = PCA(n_components=n_components)

    compressed = pca.fit_transform(img_array)

    # Reconstruct image
    reconstructed = pca.inverse_transform(compressed)

    # Normalize reconstructed image
    reconstructed = np.clip(reconstructed, 0, 255).astype(np.uint8)

    # Calculate MSE
    mse = np.mean(
        (img_array - reconstructed.astype(np.float64)) ** 2
    )

    # Calculate PSNR
    if mse == 0:
        psnr = float("inf")
    else:
        psnr = 10 * np.log10((255 ** 2) / mse)

    # Calculate approximate compression ratio
    original_parameters = img_array.size
    compressed_parameters = (
        compressed.size +
        pca.components_.size +
        pca.mean_.size
    )

    compression_ratio = original_parameters / compressed_parameters

    # Explained variance
    variance = np.sum(pca.explained_variance_ratio_) * 100

    return (
        reconstructed,
        mse,
        psnr,
        compression_ratio,
        variance
    )


# ============================================================
# FRONTEND FUNCTION
# ============================================================

def process_image(image, components):

    if image is None:
        return None, "Please upload an image."

    try:
        components = int(components)

        result, mse, psnr, ratio, variance = compress_image(
            image,
            components
        )

        information = f"""
### 📊 Compression Results

**Eigenfaces / PCA Components:** {components}

**MSE:** {mse:.4f}

**PSNR:** {psnr:.2f} dB

**Compression Ratio:** {ratio:.2f}:1

**Information Retained:** {variance:.2f}%

---

### 🔍 Interpretation

- **Lower MSE** → better reconstruction
- **Higher PSNR** → better image quality
- **Higher compression ratio** → more compression
- **Higher information retained** → more original image information preserved
"""

        return result, information

    except Exception as e:
        return None, f"Error: {str(e)}"


# ============================================================
# GRADIO FRONTEND
# ============================================================

with gr.Blocks(title="Eigenfaces Image Compression") as demo:

    gr.Markdown(
        """
        # 🖼️ Eigenfaces Image Compression

        ### PCA-Based Image Compression using Python

        Upload an image and select the number of **Eigenfaces/PCA
        components** to compress and reconstruct the image.
        """
    )

    with gr.Row():

        with gr.Column():

            input_image = gr.Image(
                type="numpy",
                label="📤 Upload Original Image"
            )

            components = gr.Slider(
                minimum=5,
                maximum=200,
                value=50,
                step=5,
                label="Number of Eigenfaces / PCA Components"
            )

            compress_button = gr.Button(
                "🚀 Compress Image",
                variant="primary"
            )

        with gr.Column():

            output_image = gr.Image(
                label="📥 Reconstructed / Compressed Image"
            )

            output_info = gr.Markdown(
                label="Compression Information"
            )

    compress_button.click(
        fn=process_image,
        inputs=[
            input_image,
            components
        ],
        outputs=[
            output_image,
            output_info
        ]
    )

    gr.Markdown(
        """
        ---
        ### 🧠 How it works

        **Original Image → Grayscale → PCA/Eigenfaces →
        Select Components → Reconstruction → Compressed Image**

        PCA finds the most important patterns in the image.
        By keeping only a limited number of components, we can
        represent the image using fewer values while preserving
        important visual information.
        """
    )


# Launch application
demo.launch(share=True)
```
